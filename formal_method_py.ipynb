```python

# ============================================================
# CASE STUDY: UNIVERSITY EXAMINATION MANAGEMENT SYSTEM
# SUBJECT: SOFTWARE ENGINEERING - FORMAL METHODS
# LANGUAGE: PYTHON
# ============================================================

print("=" * 65)
print(" UNIVERSITY EXAMINATION MANAGEMENT SYSTEM")
print(" Demonstration of Formal Methods using Python")
print("=" * 65)

# ------------------------------------------------------------
# STEP 1: DEFINE STUDENTS AND COURSES USING SETS
# Mathematical Concept: SETS
# ------------------------------------------------------------

students = {"S101", "S102", "S103", "S104", "S105"}

courses = {"Python", "Java", "Machine Learning"}

print("\n1. STUDENT AND COURSE INFORMATION")
print("Students:", students)
print("Courses:", courses)


# ------------------------------------------------------------
# STEP 2: DEFINE STUDENT ENROLMENTS
# Mathematical Concept: RELATIONS
# A relation connects students with courses.
# ------------------------------------------------------------

enrolments = {
    ("S101", "Python"),
    ("S102", "Java"),
    ("S103", "Python"),
    ("S104", "Machine Learning"),
    ("S105", "Java")
}

print("\n2. STUDENT COURSE ENROLMENTS")
for student, course in enrolments:
    print(student, "is enrolled in", course)


# ------------------------------------------------------------
# STEP 3: DEFINE ATTENDANCE RECORDS
# Mathematical Concept: FUNCTIONS
# Each student is mapped to an attendance percentage.
# ------------------------------------------------------------

attendance = {
    "S101": 85,
    "S102": 90,
    "S103": 65,
    "S104": 78,
    "S105": 72
}

print("\n3. ATTENDANCE RECORDS")
for student, percentage in attendance.items():
    print(student, ":", percentage, "%")


# ------------------------------------------------------------
# STEP 4: DEFINE FORMAL SYSTEM CONSTRAINTS
# Mathematical Concept: LOGIC AND CONDITIONS
# ------------------------------------------------------------

MIN_ATTENDANCE = 75
PASS_MARKS = 40
MAX_MARKS = 100

print("\n4. SYSTEM RULES")
print("Minimum attendance:", MIN_ATTENDANCE, "%")
print("Minimum passing marks:", PASS_MARKS)
print("Maximum marks:", MAX_MARKS)


# ------------------------------------------------------------
# STEP 5: EXAMINATION REGISTRATION
# Formal Specification:
# Preconditions:
# 1. Student must exist.
# 2. Course must exist.
# 3. Student must be enrolled.
# 4. Attendance must be at least 75%.
# 5. Duplicate registration is not allowed.
#
# Postcondition:
# Eligible student is registered successfully.
# ------------------------------------------------------------

exam_registrations = set()

def register_exam(student, course):

    if student not in students:
        return "Rejected: Student does not exist."

    if course not in courses:
        return "Rejected: Course does not exist."

    if (student, course) not in enrolments:
        return "Rejected: Student is not enrolled in this course."

    if attendance[student] < MIN_ATTENDANCE:
        return "Rejected: Attendance is below 75%."

    if (student, course) in exam_registrations:
        return "Rejected: Duplicate examination registration."

    exam_registrations.add((student, course))

    return "Success: Examination registration completed."


# ------------------------------------------------------------
# STEP 6: REGISTER STUDENTS FOR EXAMINATIONS
# ------------------------------------------------------------

print("\n5. EXAMINATION REGISTRATION RESULTS")

test_registrations = [
    ("S101", "Python"),
    ("S102", "Java"),
    ("S103", "Python"),
    ("S104", "Machine Learning"),
    ("S105", "Java"),
    ("S101", "Java"),
    ("S101", "Python")
]

for student, course in test_registrations:
    result = register_exam(student, course)
    print(student, "-", course, ":", result)


# ------------------------------------------------------------
# STEP 7: ENTER EXAMINATION MARKS
# Formal Specification:
# Marks must be between 0 and 100.
# Marks can only be entered for registered students.
# ------------------------------------------------------------

exam_marks = {}

def enter_marks(student, course, marks):

    if (student, course) not in exam_registrations:
        return "Rejected: Student is not registered for examination."

    if not isinstance(marks, (int, float)) or isinstance(marks, bool):
        return "Rejected: Marks must be numeric."

    if marks < 0 or marks > MAX_MARKS:
        return "Rejected: Marks must be between 0 and 100."

    exam_marks[(student, course)] = marks

    return "Success: Marks entered."


# ------------------------------------------------------------
# STEP 8: ENTER SAMPLE MARKS
# ------------------------------------------------------------

print("\n6. MARKS ENTRY")

marks_records = [
    ("S101", "Python", 85),
    ("S102", "Java", 67),
    ("S104", "Machine Learning", 92),
    ("S101", "Python", 105),
    ("S103", "Python", 55)
]

for student, course, marks in marks_records:
    result = enter_marks(student, course, marks)
    print(student, "-", course, "-", marks, ":", result)


# ------------------------------------------------------------
# STEP 9: GENERATE EXAMINATION RESULTS
# Mathematical Concept: FUNCTIONS
# A function maps marks to a result.
#
# Pass condition: Marks >= 40
# ------------------------------------------------------------

def calculate_result(marks):

    if marks >= PASS_MARKS:
        return "PASS"
    else:
        return "FAIL"


def generate_result(student, course):

    if (student, course) not in exam_registrations:
        return "Result unavailable: Not registered."

    if (student, course) not in exam_marks:
        return "Result unavailable: Marks not entered."

    marks = exam_marks[(student, course)]

    return calculate_result(marks)


# ------------------------------------------------------------
# STEP 10: DISPLAY FINAL RESULTS
# ------------------------------------------------------------

print("\n7. FINAL EXAMINATION RESULTS")
print("-" * 65)

for (student, course), marks in exam_marks.items():

    result = generate_result(student, course)

    print("Student ID:", student)
    print("Course:", course)
    print("Marks:", marks)
    print("Result:", result)
    print("-" * 65)


# ------------------------------------------------------------
# STEP 11: VERIFY SYSTEM RULES
# Mathematical Concept: FORMAL VERIFICATION
# Assertions check whether system constraints hold.
# ------------------------------------------------------------

print("\n8. FORMAL VERIFICATION")

# Rule 1: Every registered student must be enrolled.
assert all(
    registration in enrolments
    for registration in exam_registrations
), "Verification failed: Invalid enrolment!"

print("PASS: All registered students are enrolled.")

# Rule 2: Every registered student must meet attendance criteria.
assert all(
    attendance[student] >= MIN_ATTENDANCE
    for student, course in exam_registrations
), "Verification failed: Attendance rule violated!"

print("PASS: Attendance requirement satisfied.")

# Rule 3: All recorded marks must be between 0 and 100.
assert all(
    0 <= marks <= MAX_MARKS
    for marks in exam_marks.values()
), "Verification failed: Invalid marks!"

print("PASS: All marks are within the valid range.")

# Rule 4: Every marks record must belong to a registered student.
assert all(
    record in exam_registrations
    for record in exam_marks
), "Verification failed: Unregistered marks record!"

print("PASS: Marks belong to registered students.")

# Rule 5: No duplicate registrations.
assert len(exam_registrations) == len(set(exam_registrations))

print("PASS: No duplicate registrations.")

print("\n" + "=" * 65)
print(" ALL FORMAL VERIFICATION CHECKS COMPLETED SUCCESSFULLY!")
print("=" * 65)
```
